# 🎯 Gap Analysis Demo

This notebook demonstrates the complete curriculum gap analysis pipeline:
1. **Input:** Student skills
2. **ML Prediction:** Predicted job role
3. **Gap Analysis:** Missing skills for the predicted role

```python
import os
import sys

# Setup paths
BASE_DIR = os.path.dirname(os.getcwd())
sys.path.append(BASE_DIR)
sys.path.append(os.path.join(BASE_DIR, 'backend', 'ml_engine'))

from predictor import MLPredictor
from graph_logic import KnowledgeEngine

# Initialize components
predictor = MLPredictor()
knowledge_engine = KnowledgeEngine()

print("✅ System initialized successfully!")
```

## 🔧 Gap Analysis Function

```python
def analyze_curriculum_gap(student_skills):
    """
    Complete pipeline: Skills -> Predicted Role -> Gap Analysis
    """
    print("=" * 60)
    print("📚 CURRICULUM GAP ANALYSIS")
    print("=" * 60)
    
    # Step 1: Display input
    print(f"\n📝 INPUT SKILLS:")
    for skill in student_skills:
        print(f"   • {skill}")
    
    # Step 2: ML Prediction
    predicted_role, confidence = predictor.predict_role(student_skills)
    print(f"\n🤖 ML PREDICTION:")
    print(f"   Role: {predicted_role}")
    print(f"   Confidence: {confidence*100:.2f}%")
    
    # Step 3: Gap Analysis
    result = knowledge_engine.analyze_gap(predicted_role, student_skills)
    
    print(f"\n📊 GAP ANALYSIS:")
    print(f"   Match Score: {result['match_score']}%")
    print(f"   Skills Required: {result['total_required']}")
    print(f"   Skills Present: {len(result['present_skills'])}")
    
    print(f"\n✅ PRESENT SKILLS:")
    if result['present_skills']:
        for skill in result['present_skills']:
            print(f"   ✓ {skill}")
    else:
        print("   (None matching)")
    
    print(f"\n❌ MISSING SKILLS:")
    if result['missing_skills']:
        for skill in result['missing_skills']:
            print(f"   ✗ {skill}")
    else:
        print("   (None - Perfect match!)")
    
    print(f"\n💡 ADVICE:")
    print(f"   {result['advice']}")
    print("=" * 60)
    
    return result
```

## 🧪 Test Case 1: Web Developer

```python
skills_web = ["HTML", "CSS", "JavaScript", "React", "Git"]
analyze_curriculum_gap(skills_web)
```

## 📈 Batch Analysis

```python
# Analyze multiple skill sets
test_cases = [
    ["HTML", "CSS"],
    ["Python", "Machine Learning", "TensorFlow", "Deep Learning"],
    ["Figma", "Photoshop", "UX Design"],
    ["AWS", "Docker", "Kubernetes"],
    ["Project Management", "Agile", "Scrum"]
]

print("\n📊 BATCH ANALYSIS SUMMARY")
print("=" * 70)
print(f"{'Skills':<40} {'Predicted Role':<20} {'Confidence':>10}")
print("-" * 70)

for skills in test_cases:
    role, conf = predictor.predict_role(skills)
    skills_str = ', '.join(skills[:3]) + ('...' if len(skills) > 3 else '')
    print(f"{skills_str:<40} {role:<20} {conf*100:>8.1f}%")

print("=" * 70)
```
